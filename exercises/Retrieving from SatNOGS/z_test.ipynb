# Reading Data from SatNOGS

```python
import time
import requests

from dotenv import load_dotenv
import os

load_dotenv()
TOKEN = os.environ.get("SATNOGS_TOKEN")
```

```python
r = requests.get(
    "https://network.satnogs.org/api/observations/",
    params={"status": "good"},
)

print(r.url)          # what was actually requested
print(r.status_code)
print({o["status"] for o in r.json()})   # which statuses came back

print()

print(r.status_code)   # 200 means success
data = r.json()        # turns the reply into Python lists/dicts
print(len(data))
print(data[0].keys())  # shows you what fields each observation has
```
